```python
# Cell 1: Imports and Setup
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, classification_report

# Set the style for better visualizations
plt.style.use('seaborn')
sns.set_palette('husl')

# Custom color palette for consistency
ONSOT_COLORS = {
    'primary': '#2E86C1',    # Main blue
    'secondary': '#27AE60',  # Green
    'accent': '#E67E22',     # Orange
    'neutral': '#95A5A6',    # Gray
    'highlight': '#E74C3C'   # Red for emphasis
}
```

```python
# Cell 2: Model Accuracy Comparison Function
def plot_model_accuracy_comparison():
    # Example data - replace with actual model metrics
    models = ['OnSpot ML', 'Historical Average', 'Time-Series Baseline']
    accuracy = [0.85, 0.65, 0.70]  # Example accuracies
    
    fig = go.Figure()
    fig.add_trace(go.Bar(
        x=models,
        y=accuracy,
        text=[f'{acc:.1%}' for acc in accuracy],
        textposition='auto',
        marker_color=[ONSOT_COLORS['primary'], ONSOT_COLORS['neutral'], ONSOT_COLORS['neutral']]
    ))
    
    fig.update_layout(
        title='Model Accuracy Comparison',
        xaxis_title='Model Type',
        yaxis_title='Accuracy',
        yaxis_tickformat='.0%',
        template='plotly_white'
    )
    
    return fig

# Generate and display the plot
fig = plot_model_accuracy_comparison()
fig.show()
```

```python
# Cell 3: Predicted vs Actual Availability Function
def plot_predicted_vs_actual():
    # Example data - replace with actual predictions
    dates = pd.date_range(start='2023-01-01', periods=24, freq='H')
    actual = np.random.normal(0.7, 0.1, 24)  # Example actual values
    predicted = actual + np.random.normal(0, 0.05, 24)  # Example predictions
    
    fig = go.Figure()
    
    # Add actual values
    fig.add_trace(go.Scatter(
        x=dates,
        y=actual,
        name='Actual',
        line=dict(color=ONSOT_COLORS['primary'])
    ))
    
    # Add predictions
    fig.add_trace(go.Scatter(
        x=dates,
        y=predicted,
        name='Predicted',
        line=dict(color=ONSOT_COLORS['accent'], dash='dash')
    ))
    
    fig.update_layout(
        title='Predicted vs Actual Parking Availability',
        xaxis_title='Time',
        yaxis_title='Availability',
        yaxis_tickformat='.0%',
        template='plotly_white'
    )
    
    return fig

# Generate and display the plot
fig = plot_predicted_vs_actual()
fig.show()
```

```python
# Cell 4: Feature Importance Function
def plot_feature_importance():
    # Example data - replace with actual feature importances
    features = ['Time of Day', 'Day of Week', 'Nearby Events', 'Weather', 'Historical Trend']
    importance = [0.35, 0.25, 0.20, 0.15, 0.05]
    
    fig = go.Figure()
    fig.add_trace(go.Bar(
        x=importance,
        y=features,
        orientation='h',
        marker_color=ONSOT_COLORS['primary']
    ))
    
    fig.update_layout(
        title='Key Factors Influencing Parking Predictions',
        xaxis_title='Relative Importance',
        yaxis_title='Feature',
        xaxis_tickformat='.0%',
        template='plotly_white'
    )
    
    return fig

# Generate and display the plot
fig = plot_feature_importance()
fig.show()
```

```python
# Cell 5: Impact Metrics Function
def plot_impact_metrics():
    # Example data - replace with actual impact metrics
    metrics = {
        'Search Time Reduction': 40,
        'Emissions Reduction': 30,
        'User Satisfaction': 85
    }
    
    fig = go.Figure()
    
    for i, (metric, value) in enumerate(metrics.items()):
        fig.add_trace(go.Indicator(
            mode='gauge+number',
            value=value,
            title={'text': metric},
            gauge={'axis': {'range': [0, 100]},
                   'bar': {'color': list(ONSOT_COLORS.values())[i]}}
        ))
    
    fig.update_layout(
        grid={'rows': 1, 'columns': 3},
        template='plotly_white'
    )
    
    return fig

# Generate and display the plot
fig = plot_impact_metrics()
fig.show()
```

```python
# Cell 6: Export Functions
import os

# Create directory for exported visualizations
os.makedirs('../../../reports/visualizations/stakeholder_dashboards', exist_ok=True)

# Export the plots as HTML (interactive) and PNG (static)
def export_visualization(fig, filename):
    # Export as interactive HTML
    fig.write_html(f"../../../reports/visualizations/stakeholder_dashboards/{filename}.html")
    # Export as static PNG
    fig.write_image(f"../../../reports/visualizations/stakeholder_dashboards/{filename}.png")

# Export all visualizations
export_visualization(plot_model_accuracy_comparison(), 'model_accuracy')
export_visualization(plot_predicted_vs_actual(), 'predicted_vs_actual')
export_visualization(plot_feature_importance(), 'feature_importance')
export_visualization(plot_impact_metrics(), 'impact_metrics')
```
